## Biological Background and Study Notes

### What Is a Mismatch?

Two sequences have a mismatch at position `i` when their symbols at that position are different.

For example:

`p = GGGCCGTTGGT`

`q = GGACCGTTGAC`

Comparing the two sequences position by position reveals three mismatches.

Therefore:

`HammingDistance(p, q) = 3`

### What Is Hamming Distance?

The Hamming distance between two strings or vectors of equal length is the number of positions at which their corresponding symbols differ.

It measures the minimum number of substitutions required to transform one string into the other, assuming that only substitutions are allowed.

### Relationship to Sequence Comparison

In bioinformatics, Hamming distance provides a simple way to quantify differences between aligned sequences of equal length.

It can be used in:

- DNA sequence comparison
- Approximate pattern matching
- K-mer analysis
- Identification of sequence mismatches

### Hamming Distance and Edit Distance

Hamming distance is one of several measures used to quantify differences between sequences.

Unlike more general edit-distance measures, such as Levenshtein distance, Hamming distance considers only substitutions at corresponding positions.

Insertions and deletions are not included.

For this reason, Hamming distance is defined for sequences of equal length.

### Important Consideration

Hamming distance measures sequence differences, but does not determine their biological significance.

Two sequences with the same Hamming distance may contain mutations with very different biological consequences.

```python

def HammingDistance(p, q):
    """
    Calculate the Hamming distance between two sequences.
    The Hamming distance is the number of positions
    at which two equal-length strings differ.

    Parameters
    ----------
    p : str
        First sequence.
    q : str
        Second sequence.

    Returns
    -------
    int
        Number of mismatched positions.

    Raises
    ------
    ValueError
        If the inputs are not strings or have different lengths.
    """
    if not isinstance(p, str) or not isinstance(q, str):
        raise ValueError("Both sequences must be strings.")
    if len(p) != len(q):
        raise ValueError("Sequences must have the same length.")

    distance = 0

    for i in range(len(p)):
        if p[i] != q[i]:
            distance += 1
    return distance


# Example: Compare two DNA sequences
p = "CAGAAAGGAAGGTCCCCATACACCGACGCACCAGTTTA"
q = "CACGCCGTATGCATAAACGAGCCGCACGAACCAGAGAG"

print("Hamming distance:", HammingDistance(p, q))
```

### Implementation Notes

The algorithm compares two sequences position by position and increments a counter whenever the symbols differ.

For sequences of length `n`, the time complexity is O(n).

The implementation is intended for educational use and demonstrates a foundational approach to comparing biological sequences.

```python
# Test 1: Basic Hamming distance
assert HammingDistance("GGGCCGTTGGT", "GGACCGTTGAC") == 3
print("Test 1 passed: Basic Hamming distance")

# Test 2: Identical sequences
assert HammingDistance("ACGTACGT", "ACGTACGT") == 0
print("Test 2 passed: Identical sequences")

# Test 3: Completely different sequences
assert HammingDistance("AAAA", "TTTT") == 4
print("Test 3 passed: Completely different sequences")

# Test 4: Single nucleotide mismatch
assert HammingDistance("ACGT", "ACGA") == 1
print("Test 4 passed: Single nucleotide mismatch")

# Test 5: Empty sequences
assert HammingDistance("", "") == 0
print("Test 5 passed: Empty sequences")

# Test 6: Reject sequences with different lengths
try:
    HammingDistance("ACGT", "ACG")
except ValueError:
    print("Test 6 passed: Different sequence lengths rejected")
else:
    raise AssertionError("Different sequence lengths were accepted")

# Test 7: Reject non-string inputs
invalid_inputs = [(123, "ACG"), ("ACG", None), (["A", "C"], "AC")]

for p, q in invalid_inputs:
    try:
        HammingDistance(p, q)
    except ValueError:
        continue

    raise AssertionError(f"Invalid input was accepted: {p}, {q}")
print("Test 7 passed: Non-string inputs rejected")
```
