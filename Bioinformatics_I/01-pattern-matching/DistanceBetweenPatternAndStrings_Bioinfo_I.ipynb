# Distance Between Pattern and Strings

**Topic:** DNA Sequence Analysis and Motif Discovery

### Overview

This notebook implements an algorithm to calculate the total distance between a nucleotide pattern and a collection of DNA sequences.

For each DNA sequence, the algorithm identifies the k-mer with the smallest Hamming distance to the given pattern.

The total distance is obtained by summing these minimum distances across all sequences.

- Calculate the minimum mismatch distance for each sequence.

### Problem Definition

**Input:** A nucleotide pattern `Pattern` and a collection of DNA sequences `Dna`.

**Output:** The total distance `d(Pattern, Dna)`.

## Algorithm and Study Notes

### Distance Between a Pattern and DNA Strings

The distance between a pattern and a collection of DNA sequences is calculated by finding the most similar k-mer in each sequence.

For every DNA string, the algorithm evaluates all possible substrings of length `k`, where `k` is the length of the pattern.

The minimum Hamming distance is selected for each sequence, and these values are added to obtain the total distance.

### Pseudocode

```text
DistanceBetweenPatternAndStrings(Pattern, Dna)
    k ← length(Pattern)
    distance ← 0

    for each string Text in Dna
        minimum_distance ← infinity

        for each k-mer PatternPrime in Text
            current_distance ← HammingDistance(Pattern, PatternPrime)

            if current_distance < minimum_distance
                minimum_distance ← current_distance

        distance ← distance + minimum_distance

    return distance
```

### Step-by-Step Explanation

1. Determine the length `k` of the target pattern.
2. Initialize the total distance to zero.
3. For each DNA sequence, initialize its minimum distance to infinity.
4. Extract every possible k-mer from the sequence.
5. Calculate the Hamming distance between the target pattern and each k-mer.
6. Retain the smallest Hamming distance found in that sequence.
7. Add this minimum distance to the total.
8. Return the sum after processing all DNA sequences.

### Relationship to Motif Discovery

This algorithm is relevant to computational motif discovery.

A candidate motif may not appear identically in every DNA sequence, but similar k-mers may be present.

The total pattern-to-strings distance provides a measure of how closely a candidate motif resembles the best-matching regions across a collection of sequences.

A smaller total distance indicates that the pattern has closer matches within the analyzed sequences.

### Hamming Distance

Hamming distance counts the number of mismatched positions between two equal-length sequences.

For example:

```python
HammingDistance("AAA", "AAT")
```

Returns:

```text
1
```

The algorithm evaluates substitutions only and does not account for insertions or deletions.

````python
from pathlib import Path

def HammingDistance(p, q):
    """
    Calculate the Hamming distance between two equal-length strings.

    Parameters
    ----------
    p, q : str
        Sequences to compare.

    Returns
    -------
    int
        Number of mismatched positions.

    Raises
    ------
    ValueError
        If the inputs are not strings or have different lengths.
    """
    if not isinstance(p, str) or not isinstance(q, str):
        raise ValueError("Both sequences must be strings.")
    if len(p) != len(q):
        raise ValueError("Sequences must have the same length.")

    distance = 0

    for i in range(len(p)):
        if p[i] != q[i]:
            distance += 1
    return distance


def Distance_BetweenPatternAndStrings(pattern, dna):
    """
    Calculate the total distance between a pattern and DNA strings.
    For each DNA sequence, find the k-mer with the minimum
    Hamming distance to the pattern.

    Return the sum of these minimum distances.

    Parameters
    ----------
    pattern : str
        Nucleotide pattern to compare.
    dna : list[str]
        Collection of DNA sequences.

    Returns
    -------
    int
        Sum of the minimum Hamming distances.

    Raises
    ------
    ValueError
        If the pattern or DNA collection is invalid.
    """
    if not isinstance(pattern, str) or not pattern:
        raise ValueError("Pattern must be a non-empty string.")
    if not isinstance(dna, (list, tuple)) or not dna:
        raise ValueError("DNA must be a non-empty collection of sequences.")

    k = len(pattern)
    distance = 0

    for text in dna:
        if not isinstance(text, str):
            raise ValueError("All DNA sequences must be strings.")
        if len(text) < k:
            raise ValueError(
                "Each DNA sequence must be at least as long as the pattern."
            )

        minimum_distance = float("inf")

        for i in range(len(text) - k + 1):
            pattern_prime = text[i:i + k]
            current_distance = HammingDistance(pattern, pattern_prime)
            if current_distance < minimum_distance:
                minimum_distance = current_distance
        distance += minimum_distance
    return distance


def load_pattern_and_dna(dataset_path):
    """
    Load a nucleotide pattern and DNA sequences from a text file.
    The first line contains the pattern.
    The second line contains space-separated DNA sequences.

    Parameters
    ----------
    dataset_path : str or Path
        Path to the input dataset.

    Returns
    -------
    tuple[str, list[str]]
        Pattern and DNA sequence collection.

    Raises
    ------
    ValueError
        If the dataset is incomplete or contains empty inputs.
    """
    with Path(dataset_path).open("r", encoding="utf-8") as file:
        lines = file.readlines()

    if len(lines) < 2:
        raise ValueError(
            "Dataset must contain a pattern and DNA sequences."
        )

    pattern = lines[0].strip()
    dna = lines[1].split()

    if not pattern or not dna:
        raise ValueError(
            "Dataset must contain non-empty pattern and DNA sequences."
        )
    return pattern, dna


# Load the original course dataset
dataset_path = Path("../dataset/dataset_30312_1.txt")

pattern, dna = load_pattern_and_dna(dataset_path)

# Calculate the total distance between the pattern and DNA sequences
distance = Distance_BetweenPatternAndStrings(pattern, dna)

# Display the results
print("Pattern:", pattern)
print("Number of DNA sequences:", len(dna))
print("Total distance:", distance)
````

### Implementation Notes

For each DNA sequence, the algorithm examines all possible k-mers of the same length as the target pattern.

The Hamming distance is calculated for each candidate, and the minimum value contributes to the final result.

For `t` DNA sequences of length approximately `n` and a pattern of length `k`, the worst-case time complexity is O(tnk).

The implementation is intended for educational use and demonstrates a computational approach relevant to motif discovery.
