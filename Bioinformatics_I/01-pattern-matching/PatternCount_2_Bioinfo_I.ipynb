## Biological Background and Study Notes

### Searching for Two DNA Patterns

The `PatternCount_2()` function counts the number of exact occurrences of a nucleotide pattern within a DNA sequence.

To compare two different patterns, the function is applied separately to each one.

In this example, the target patterns are:

- `ACCAC`
- `TACCA`

Both patterns contain five nucleotides and are therefore classified as 5-mers.

### Biological Context: oriC

The origin of chromosomal replication, commonly referred to as `oriC` in bacteria, is a genomic region associated with the initiation of DNA replication.

Bacterial replication origins contain sequence features that contribute to the recognition and initiation of DNA replication.

In the original coursework, a DNA sequence identified as an oriC region from *T. petrophila* is examined for occurrences of two specific nucleotide patterns.

Finding frequent patterns can help investigate sequence characteristics, but occurrence counts alone cannot establish the biological function of a candidate motif.

### Counting Patterns Independently

The algorithm searches the sequence once for each target pattern.

For the first pattern:

```python
count_1 = PatternCount_2(Text, Pattern_1)
```

For the second pattern:

```python
count_2 = PatternCount_2(Text, Pattern_2)
```

The total number of occurrences is calculated as:

```python
total = count_1 + count_2
```

This total represents the sum of the individual pattern counts.

### Results

| Pattern | Exact occurrences |
|---|---:|
| ACCAC | 6 |
| TACCA | 6 |
| **Total** | **12** |

Both patterns occur six times in the analyzed sequence.

### Important Considerations

- Exact pattern counting does not allow nucleotide mismatches.
- Occurrences are counted independently for each target pattern.
- Overlapping occurrences of the same pattern can be counted.
- The reported total is the sum of the two individual frequencies.
- Pattern frequency alone does not establish biological significance.

````python

def PatternCount_2(Text, Pattern):
    """
    Count exact occurrences of a pattern in a DNA sequence.
    Overlapping occurrences are included.

    Parameters
    ----------
    Text : str
        DNA sequence to analyze.
    Pattern : str
        Nucleotide pattern to count.

    Returns
    -------
    int
        Number of exact pattern occurrences.

    Raises
    ------
    ValueError
        If the inputs are not strings or the pattern is empty.
    """
    if not isinstance(Text, str) or not isinstance(Pattern, str):
        raise ValueError("Text and Pattern must be strings.")
    if not Pattern:
        raise ValueError("Pattern cannot be empty.")

    count = 0
    k = len(Pattern)

    for i in range(len(Text) - k + 1):
        if Text[i:i + k] == Pattern:
            count += 1
    return count


Pattern_1 = "ACCAC"
Pattern_2 = "TACCA"

# oriC region from T. petrophila
Text = "AACTCTATACCTCCTTTTTGTCGAATTTGTGTGATTTATAGAGAAAATCTTATTAACTGAAACTAAAATGGTAGGTTTGGTGGTAGGTTTTGTGTACATTTTGTAGTATCTGATTTTTAATTACATACCGTATATTGTATTAAATTGACGAACAATTGCATGGAATTGAATATATGCAAAACAAACCTACCACCAAACTCTGTATTGACCATTTTAGGACAACTTCAGGGTGGTAGGTTTCTGAAGCTCTCATCAATAGACTATTTTAGTCTTTACAAACAATATTACCGTTCAGATTCAAGATTCTACAACGCTGTTTTAATGGGCGTTGCAGAAAACTTACCACCTAAAATCCAGTATCCAAGCCGATTTCAGAGAAACCTACCACTTACCTACCACTTACCTACCACCCGGGTGGTAAGTTGCAGACATTATTAAAAACCTCATCAGAAGCTTGTTCAAAAATTTCAATACTCGAAACCTACCACCTGCGTCCCCTATTATTTACTACTACTAATAATAGCAGTATAATTGATCTGA"

count_1 = PatternCount_2(Text, Pattern_1)
count_2 = PatternCount_2(Text, Pattern_2)
total = count_1 + count_2

# Display individual pattern counts and their sum
print(f"Pattern 1 ({Pattern_1}): {count_1}")
print(f"Pattern 2 ({Pattern_2}): {count_2}")
print(f"Total occurrences: {total}")
````

### Implementation Notes

The algorithm uses Python string slicing and a sliding window.

For a DNA sequence of length `n` and a pattern of length `k`, a single pattern-counting operation has a worst-case time complexity of O(nk).

This notebook demonstrates the application of a basic string-processing algorithm to a biological sequence example.

````python
# Test 1: Basic pattern counting
assert PatternCount_2("GACCATACTG", "ATA") == 1
print("Test 1 passed: Basic pattern counting")

# Test 2: Overlapping occurrences
assert PatternCount_2("AAAAA", "AAA") == 3
print("Test 2 passed: Overlapping occurrences")

# Test 3: Pattern not found
assert PatternCount_2("ACGTACGT", "TTT") == 0
print("Test 3 passed: Pattern not found")

# Test 4: Two patterns in the same sequence
test_text = "ATGATGCCC"
test_count_1 = PatternCount_2(test_text, "ATG")
test_count_2 = PatternCount_2(test_text, "CCC")

assert test_count_1 == 2
assert test_count_2 == 1
assert test_count_1 + test_count_2 == 3
print("Test 4 passed: Two patterns counted independently")

# Test 5: Pattern longer than sequence
assert PatternCount_2("ACG", "ACGT") == 0
print("Test 5 passed: Pattern longer than sequence")

# Test 6: Reject invalid inputs
invalid_inputs = [("ACGT", ""), (123, "AC"), ("ACGT", None), (None, "AC")]

for text, pattern in invalid_inputs:
    try:
        PatternCount_2(text, pattern)
    except ValueError:
        continue
    raise AssertionError(f"Invalid input was accepted: {text}, {pattern}")
print("Test 6 passed: Invalid inputs rejected")


# Test 7: Verify counts in the original oriC example
oric_count_1 = PatternCount_2(Text, Pattern_1)
oric_count_2 = PatternCount_2(Text, Pattern_2)
oric_total = oric_count_1 + oric_count_2

assert oric_count_1 == Text.count(Pattern_1)
assert oric_count_2 == Text.count(Pattern_2)
assert oric_total == Text.count(Pattern_1) + Text.count(Pattern_2)

print("Test 7 passed: Original oriC example verified")
print("Pattern 1 count:", oric_count_1)
print("Pattern 2 count:", oric_count_2)
print("Total occurrences:", oric_total)
````
