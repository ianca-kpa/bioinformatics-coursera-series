## Biological Background and Study Notes

### How Can We Have a Reference Genome If Mutations Occur During DNA Replication?

Mutations are an important source of genetic variation and contribute to evolution across all domains of life.

Although mutations can arise during DNA replication, the mutation rate is relatively low compared with the size of a genome. Consequently, a reference genome can provide a useful representation of a species or individual, even though genetic variation exists.

A reference genome is not a perfectly fixed or universal sequence. Rather, it serves as a standardized representation against which genetic differences can be studied.

### Germline Mutations and Genetic Variation

Some mutations arise in reproductive cells or their precursors and can be transmitted to offspring.

De novo mutations are genetic variants found in an individual that were not detected in the parental genomes.

Human genomes typically acquire dozens of new single-nucleotide variants per generation, with estimates varying according to the methods and populations studied.

The number of de novo point mutations is associated with parental age, particularly paternal age, although maternal age can also influence mutation patterns.

Genetic differences accumulate between individuals and populations. Approximately 0.1% of nucleotide positions are commonly cited as differing between two human genomes when considering single-nucleotide variation, although the overall variation depends on the types of genomic differences included.

### Somatic Mutations

Mutations also accumulate in somatic cells throughout an individual's lifetime.

These changes can result from DNA replication errors, spontaneous chemical modifications, environmental exposures, and other cellular processes.

Most somatic mutations do not cause cancer. However, certain mutations affecting genes involved in cell growth, DNA repair, or cell-cycle regulation can contribute to tumor development.

Somatic mutations also explain why genetically distinct cell populations may arise within the same individual.

### DNA Repair Mechanisms

Cells have mechanisms that detect and repair DNA damage and replication errors.

These mechanisms help preserve genomic stability and reduce the accumulation of harmful mutations.

Examples include:

- DNA polymerase proofreading
- Mismatch repair
- Base excision repair
- Nucleotide excision repair

These pathways act on different types of DNA errors or damage.

### Cytosine Deamination

Cytosine can undergo spontaneous deamination, producing uracil.

Because uracil is normally found in RNA rather than DNA, its presence in DNA can indicate damage.

Uracil-DNA glycosylase participates in base excision repair by removing uracil from DNA, allowing subsequent repair steps to restore the appropriate nucleotide.

When 5-methylcytosine undergoes deamination, it produces thymine instead of uracil.

This can create a T:G mismatch, which may be recognized by repair proteins, including thymine-DNA glycosylase in appropriate sequence contexts.

If these changes are not repaired before replication, they can become permanent mutations.

### Why Do Mutations Persist?

DNA repair is highly effective but not perfect.

Some genetic changes escape repair and become established during subsequent DNA replication.

Mutations that are compatible with cell survival may persist, and some can be passed to daughter cells.

In reproductive lineages, inherited mutations contribute to genetic variation across generations.

### Relevance to Bioinformatics

Advances in DNA sequencing and computational analysis make it possible to identify genetic variation between individuals, tissues, and cell populations.

Bioinformatics approaches help researchers detect sequence differences, compare genomes, and investigate the biological consequences of mutations.

## Understanding k-mers and Python Slicing

### What Is a k-mer?

A **k-mer** is a contiguous subsequence of length `k` within a longer biological sequence.

For example, consider the DNA sequence:

`GACCATACTG`

Its 3-mers are:

`GAC`, `ACC`, `CCA`, `CAT`, `ATA`, `TAC`, `ACT`, `CTG`

The length of a k-mer is determined by the value of `k`.

### Applications of k-mers in Bioinformatics

**1. Motif Discovery**

K-mers can represent candidate sequence motifs, which may correspond to biologically meaningful patterns such as protein-binding sites.

**2. Sequence Similarity**

Comparing k-mer composition between biological sequences can help identify shared sequence features and estimate similarity.

**3. Genome Assembly**

Graph-based genome assembly methods use relationships between overlapping k-mers to help reconstruct longer sequences from shorter fragments.

**4. Frequency Analysis**

Counting k-mer occurrences provides information about nucleotide composition and the distribution of repeated sequence patterns.

### What Is Python Slicing?

Python slicing extracts a portion of a sequence, such as a string, list, or tuple.

The general syntax is:

```python
sequence[start:end]
```

- `start`: Index of the first element included in the slice.
- `end`: Index immediately after the final included element.

The ending index is exclusive.

For example:

```python
text = "GACCATACTG"
print(text[2:5])
```

Output:

```text
CCA
```

If `start` is omitted, slicing begins at index `0`.

If `end` is omitted, slicing continues to the end of the sequence.

### Sliding Windows and Pattern Matching

The `PatternCount()` algorithm uses slicing to examine consecutive windows of length `k`.

For a sequence of length `n` and a pattern of length `k`, the final valid starting position is:

`n - k`

Therefore, the algorithm uses:

```python
for i in range(len(Text) - len(Pattern) + 1):
```

For example, in a sequence of length `10`, the final 3-mer starts at position `7`.

The `+1` is necessary because Python's `range()` excludes its upper bound.

### Overlapping Patterns

Exact pattern matching may include overlapping occurrences.

For example:

```python
PatternCount("AAAAA", "AAA")
```

Returns:

```text
3
```

The pattern occurs at starting positions `0`, `1`, and `2`.

````python
def PatternCount(Text, Pattern):
    """
    Count the number of exact occurrences of a pattern in a DNA sequence.
    Overlapping occurrences are included.

    Parameters
    ----------
    Text : str
        DNA sequence to search.
    Pattern : str
        Nucleotide pattern to count.

    Returns
    -------
    int
        Number of exact pattern occurrences.

    Raises
    ------
    ValueError
        If either input is not a string or the pattern is empty.
    """
    if not isinstance(Text, str) or not isinstance(Pattern, str):
        raise ValueError("Text and Pattern must be strings.")
    if not Pattern:
        raise ValueError("Pattern cannot be empty.")
    count = 0
    k = len(Pattern)
    for i in range(len(Text) - k + 1):
        if Text[i:i + k] == Pattern:
            count += 1
    return count


# Example: Count exact pattern occurrences
text = "TACGCATTACAAAGCACA"
pattern = "AA"

print("Pattern count:", PatternCount(text, pattern))
````

## Validation and Testing

The `PatternCount()` implementation was evaluated using six automated tests.

### Test Cases

1. **Basic pattern counting:** Verify exact pattern occurrences in a DNA sequence.
2. **Overlapping occurrences:** Confirm that overlapping matches are counted separately.
3. **Pattern not found:** Return zero when the target pattern is absent.
4. **Pattern longer than the sequence:** Return zero when the pattern is longer than the input.
5. **Exact sequence match:** Return one when the pattern matches the complete sequence.
6. **Invalid inputs:** Reject empty patterns and non-string inputs.

### Results

All six tests passed.

### Implementation Notes

The algorithm uses a sliding window and Python string slicing.

For a sequence of length `n` and a non-empty pattern of length `k`, it examines `max(0, n - k + 1)` candidate positions.

The time complexity is O(nk) in the general case because each substring comparison may involve up to `k` characters.

The implementation is intended for educational use and demonstrates a foundational string-processing technique in bioinformatics.

````python
# Test 1: Basic pattern counting
assert PatternCount("GACCATACTG", "ATA") == 1
print("Test 1 passed: Basic pattern counting")

# Test 2: Overlapping occurrences
assert PatternCount("AAAAA", "AAA") == 3
print("Test 2 passed: Overlapping occurrences")

# Test 3: Pattern not found
assert PatternCount("ACGTACGT", "TTT") == 0
print("Test 3 passed: Pattern not found")

# Test 4: Pattern longer than the sequence
assert PatternCount("ACG", "ACGT") == 0
print("Test 4 passed: Pattern longer than sequence")

# Test 5: Exact sequence match
assert PatternCount("ACGT", "ACGT") == 1
print("Test 5 passed: Exact sequence match")


# Test 6: Reject invalid inputs
invalid_inputs = [("ACGT", ""), (123, "A"), ("ACGT", 123), (None, "A")]

for text, pattern in invalid_inputs:
    try:
        PatternCount(text, pattern)
    except ValueError:
        continue
    raise AssertionError(f"Invalid input was accepted: {text}, {pattern}")
print("Test 6 passed: Invalid inputs rejected")
````
